```python
import numpy as np
import pandas as pd
import pickle
from tensorflow.keras.models import load_model
from tensorflow.keras.preprocessing.sequence import pad_sequences

# Load trained model
model = load_model("model.h5")

# Load tokenizer
with open("tokenizer.pkl", "rb") as f:
    tokenizer = pickle.load(f)

# Load config
with open("config.pkl", "rb") as f:
    config = pickle.load(f)

max_len = config["max_len"]

print("✅ Model, tokenizer, and config loaded successfully!")
print(f"Max sequence length: {max_len}")
print(f"Vocabulary size: {len(tokenizer.word_index)}")
```

```python
def generate_response(text):
    """
    Generate chatbot response for given input text.
    
    Process:
    1. Convert text to sequences using tokenizer
    2. Pad sequences to max_len
    3. Predict using trained model
    4. Convert predictions back to words
    5. Return response as string
    """
    # Convert text to sequences
    seq = tokenizer.texts_to_sequences([text])
    
    # Pad sequences
    seq = pad_sequences(seq, maxlen=max_len, padding='post')
    
    # Get predictions
    pred = model.predict(seq, verbose=0)
    
    # Get argmax (highest probability token)
    pred = np.argmax(pred, axis=-1)[0]
    
    # Convert token indices back to words
    result = []
    for idx in pred:
        for word, index in tokenizer.word_index.items():
            if index == idx:
                result.append(word)
                break
    
    return " ".join(result)

print("✅ Response generation function ready!")
```

```python
test_1 = "Where is Angkor Wat?"
response_1 = generate_response(test_1)

print("=" * 60)
print(f"TEST 1: Angkor Wat Location")
print("=" * 60)
print(f"Input:  {test_1}")
print(f"Output: {response_1}")
print()
```
